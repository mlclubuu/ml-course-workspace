```python
import pandas as pd  
```

```python
df = pd.read_csv('play_football_dataset.csv')
```

```python
df
```

```python
X = df.drop(columns='Play Football?', axis=1) 
```

```python
X = df.drop(columns=['Day', 'Play Football?'], axis=1)
```

```python
X
```

```python
y = df['Play Football?']
```

```python
y
```

```python
from sklearn.preprocessing import OrdinalEncoder
encoder = OrdinalEncoder()
X_encoded = encoder.fit_transform(X)
X_encoded
```

```python
y_encoded = y.map({ 'Yes':1 , 'No':0})
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y_encoded, test_size=0.3)
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(X_train,y_train)
```

```python
model.score(X_test, y_test)
```

```python
from joblib import load, dump
dump(model, 'decisiontree.joblib')
```
