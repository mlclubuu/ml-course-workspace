```python
df = pd.read_csv('play_football_dataset.csv')
```

```python
df.drop(['Day'], axis=1, inplace=True)
```

```python
X = df[["Weather","Temperature", "Humidity", "Wind"]]
y = df['Play Football?']
```

```python

Weather = {
    'Cloudy' : 0,
    'Rain' : 1,
    'Sunny' : 2
}
Temparature = {
    'Cool' : 0,
    'Hot' : 1,
    'Mild' : 2
}

Humidity = {
    'High' : 0,
    'Normal' : 1
}

Wind = {
    'Strong' : 0,
    'Weak' : 1
}
```

```python
def get_input_value(weather , temparature, humadity, wind):
    return [[Weather[weather], Temparature[temparature], Humidity[humadity], Wind[wind]]]
```

```python
def get_permission(permission):
    if permission == 1:
        return 'Yes you can play football'
    else:
        return "No you can't play football today"
```

```python
get_input_value('Sunny','Hot','High','Weak')
```

```python
from sklearn.preprocessing import OrdinalEncoder
encoder = OrdinalEncoder()

X_encoded = encoder.fit_transform(X)
X_encoded = pd.DataFrame(X_encoded)
X_encoded
```

```python
X_encoded[0][3]
```

```python
y_encoded = y.map({'Yes':1, 'No':0})
```

```python
y_encoded
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
```

```python
from sklearn.model_selection import train_test_split
X_train , X_test , y_train , y_test = train_test_split(X_encoded, y_encoded, test_size=0.3, random_state=42)
```

```python
model.fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```

```python

permisison =  model.predict(get_input_value('Rain','Cool','Normal','Strong'))
print(get_permission(permisison))
```
